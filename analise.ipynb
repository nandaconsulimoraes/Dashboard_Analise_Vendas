Importando Bibliotecas e Carregamento da base de dados

```python
#importar as bibliotecas
import pandas as pd

#carregar os arquivos
df_analise = pd.read_csv("Vendas_Chocolate.csv", sep =";")
```

---

Inspeção Inicial (explorando e entendendo dados)

```python
#inspeção inicial
display(df_analise.shape)
display(df_analise.head())
df_analise.info()
```

---

Tratamento de Dados

```python
#valores nulos
df_analise_tratada = df_analise.drop(columns= ["Data de Hoje"])
df_analise_tratada = df_analise_tratada.dropna()

#tipos de dados
df_analise_tratada["Valor"] = df_analise_tratada["Valor"].str.replace("$","").str.replace(",","").astype(float)
df_analise_tratada["Data"] = pd.to_datetime(df_analise_tratada["Data"], format = "%d/%m/%Y")

#valores duplicados
df_analise_tratada = df_analise_tratada.drop_duplicates()

#reset index
df_analise_tratada = df_analise_tratada.reset_index(drop = True)

display(df_analise_tratada)
```

```python
display(df_analise_tratada.describe())
```

---

Feature Engineering

```python
#feature engineering

#criar uma coluna para mes
df_analise_tratada["Mes"]= df_analise_tratada["Data"].dt.to_period("M")

#criar uma coluna para ano
df_analise_tratada["Ano"]= df_analise_tratada["Data"].dt.year
display(df_analise_tratada)
```

___

Análise de crescimento de vendas

```python
#análise temporal(dias/mes/ano)
import plotly.express as px

vendas_mes = df_analise_tratada[["Mes", "Valor"]]
vendas_mes = vendas_mes.groupby(by="Mes", as_index = False).sum()
display(vendas_mes)

grafico = px.bar(vendas_mes, x= vendas_mes["Mes"].astype(str), y ="Valor")
grafico.show()
```

```python
#análise de volume e ticket médio
vendas_ano = df_analise_tratada[["Ano", "Valor","Caixas Enviadas"]]
vendas_ano = vendas_ano.groupby(by="Ano", as_index = False).sum()

#Feature engineering(criando coluna de ticket médio e de percentual de crescimento)
vendas_ano["Ticket Médio"] = vendas_ano["Valor"] / vendas_ano["Caixas Enviadas"]
vendas_ano["% Crescimento Valor"] = vendas_ano["Valor"].pct_change()


display(vendas_ano.style.format({
    "Valor": "${:,.2f}",
    "Caixas Enviadas": "{:,.0f}",
    "Ticket Médio": "${:,.2f}",
    "% Crescimento Valor": "{:,.1%}"
}))
```

```python
#análise por produto

df_produtos_anual = df_analise_tratada[["Ano", "Produto", "Valor"]]
df_produtos_anual= df_produtos_anual.groupby(["Ano", "Produto"], as_index = False).sum()
display(df_produtos_anual)
grafico = px.bar(df_produtos_anual, x = "Ano", y = "Valor", color = "Produto")
grafico.show()
```

---

Quais produtos mais vendem?

```python
produtos_quantidade = df_analise_tratada.groupby("Produto")["Caixas Enviadas"].sum().sort_values(ascending=False).reset_index()
display(produtos_quantidade)
grafico = px.bar(produtos_quantidade, x = "Produto", y="Caixas Enviadas", title = "Quantidade de caixas vendidas por produto")
grafico.show()
```

#### Insight:
O produto "Mordidas 50% Cacau" apresentou a maior quantidade de caixas vendidas, representando o item de maior demanda no período analisado.

---

Quais produtos geram mais faturamento? 

```python
faturamento = df_analise_tratada.groupby("Produto")["Valor"].sum().sort_values(ascending = False).reset_index()
display(faturamento)
grafico = px.bar(faturamento, x ="Produto" , y = "Valor", title= "Faturamento por produto")
grafico.show()
```

#### Insight:

Apesar de o produto "Salgadinho Suave e Sedoso" vender menos unidades comparado com o produto que mais vende (Mordidas 50% Cacau), ele gera maior faturamento, indicando um preço médio mais elevado

---

Existe relação entre a quantidade vendida e faturamento?


```python
grafico = px.scatter(df_analise_tratada, x = "Caixas Enviadas", y = "Valor", color = "Produto", trendline="ols", title="Relação quantidade vendida e faturamento")
grafico.show()
```

#### Insight:

Apesar de o produto "Mordidas 50% Cacau "apresentar maior volume de caixas vendidas, o "Salgadinho Suave e Sedoso "foi o responsável pelo maior faturamento. Isso indica que o desempenho financeiro de um produto não depende apenas da quantidade vendida, mas também do seu valor por venda.

---

Existe algum pais que compra mais? 


```python
pais = df_analise_tratada.groupby("País")["Valor"].sum().sort_values(ascending = False).reset_index()
grafico = px.bar(pais, x = "País", y = "Valor", title="Países que mais realizam compras")
grafico.show()

faturamento_total = pais["Valor"].sum()
pais["Participação (%)"]= (pais["Valor"] / faturamento_total * 100).round(2)
display(pais)
```

#### Insights:
A Austrália apresentou o maior faturamento entre os países analisados, representando 18,44% do faturamento total da empresa. Esse resultado evidencia a importância desse mercado para o desempenho financeiro da organização e pode orientar decisões estratégicas relacionadas a investimentos e expansão comercial.
